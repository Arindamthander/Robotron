```python
import numpy as np

# Create a random matrix A of size 5x11
A = np.random.rand(5, 11)

# Compute SVD decomposition without using built-in function
def svd_custom(A):
    # Calculate covariance matrix
    covariance_matrix = A.T @ A

    # Eigenvalue decomposition of covariance matrix
    eigenvalues, eigenvectors = np.linalg.eigh(covariance_matrix)

    # Sort eigenvalues and eigenvectors in descending order
    order = np.argsort(eigenvalues)[::-1]
    eigenvalues = eigenvalues[order]
    eigenvectors = eigenvectors[:, order]

    # Construct Sigma matrix
    sigma = np.sqrt(np.abs(eigenvalues))

    # Construct U and V matrices

    Vt = eigenvectors.T
    U = npdot(np.dot(A,np.linalg.inverse(Vt)),np.linalg()

    return U, sigma, Vt

# Compute SVD using the custom function
U_custom, sigma_custom, Vt_custom = svd_custom(A)

# Compute SVD using the built-in function for comparison
U_np, sigma_np, Vt_np = np.linalg.svd(A, full_matrices=False)

# Compare the results
print("U (Custom vs NumPy):")
print(U_custom)
print(U_np)

print("\nSigma (Custom vs NumPy):")
print(sigma_custom)
print(sigma_np)

print("\nVt (Transpose) (Custom vs NumPy):")
print(Vt_custom)
print(Vt_np)
```
